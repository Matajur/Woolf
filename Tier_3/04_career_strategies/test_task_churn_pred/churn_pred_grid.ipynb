```python
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import matthews_corrcoef
from lightgbm import LGBMClassifier

# ============ Config ============
N_SPLITS = 5
BASE_SEED = 42

# If this is too slow, set to 3 first:
# N_SPLITS = 3


def mcc_optimal_threshold(y_true, proba, grid=None):
    if grid is None:
        grid = np.linspace(0.05, 0.95, 181)
    best_t, best_mcc = 0.5, -1.0
    for t in grid:
        mcc = matthews_corrcoef(y_true, (proba >= t).astype(int))
        if mcc > best_mcc:
            best_mcc = mcc
            best_t = float(t)
    return best_t, float(best_mcc)


def cv_eval_lgb_params(X, y, X_test, params, n_splits=5, seed=42):
    cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed)
    oof = np.zeros(len(X), dtype=float)
    test_proba = np.zeros(len(X_test), dtype=float)

    for fold, (tr_idx, va_idx) in enumerate(cv.split(X, y), 1):
        X_tr, X_va = X.iloc[tr_idx], X.iloc[va_idx]
        y_tr, y_va = y.iloc[tr_idx], y.iloc[va_idx]

        model = LGBMClassifier(**params)
        model.fit(
            X_tr,
            y_tr,
            eval_set=[(X_va, y_va)],
            eval_metric="binary_logloss",
        )

        oof[va_idx] = model.predict_proba(X_va)[:, 1]
        test_proba += model.predict_proba(X_test)[:, 1] / n_splits

    best_t, best_mcc = mcc_optimal_threshold(y, oof)
    return best_mcc, best_t, oof, test_proba


# ============ Load data ============
train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")

ID_COL = train.columns[0]  # unnamed ID column from your preview
TARGET_COL = "target_class"

test_ids = test[ID_COL].values

X = train.drop(columns=[ID_COL, TARGET_COL])
y = train[TARGET_COL].astype(int)
X_test = test.drop(columns=[ID_COL])

# ============ Small, high-signal grid ============
# Notes:
# - This grid explores capacity (leaves/depth), regularization, and sampling.
# - All use many trees with low learning rate; MCC threshold will be tuned on OOF.
base = dict(
    n_estimators=8000,
    learning_rate=0.01,
    random_state=BASE_SEED,
    n_jobs=-1,
)

param_grid = [
    # 1) Solid baseline (balanced bias/variance)
    dict(
        **base,
        num_leaves=63,
        max_depth=-1,
        min_child_samples=15,
        subsample=0.9,
        colsample_bytree=0.9,
        reg_alpha=0.1,
        reg_lambda=0.1,
    ),
    # 2) Slightly more regularized
    dict(
        **base,
        num_leaves=63,
        max_depth=-1,
        min_child_samples=25,
        subsample=0.85,
        colsample_bytree=0.85,
        reg_alpha=0.5,
        reg_lambda=0.5,
    ),
    # 3) Higher capacity, mild reg (can win if signal is strong)
    dict(
        **base,
        num_leaves=127,
        max_depth=-1,
        min_child_samples=15,
        subsample=0.9,
        colsample_bytree=0.9,
        reg_alpha=0.1,
        reg_lambda=0.1,
    ),
    # 4) Higher capacity + stronger reg (safer)
    dict(
        **base,
        num_leaves=127,
        max_depth=-1,
        min_child_samples=25,
        subsample=0.85,
        colsample_bytree=0.85,
        reg_alpha=0.5,
        reg_lambda=0.8,
    ),
    # 5) Shallower trees (sometimes improves MCC on engineered features)
    dict(
        **base,
        num_leaves=63,
        max_depth=6,
        min_child_samples=15,
        subsample=0.9,
        colsample_bytree=0.9,
        reg_alpha=0.1,
        reg_lambda=0.1,
    ),
    # 6) Shallow + strong reg (very robust)
    dict(
        **base,
        num_leaves=63,
        max_depth=6,
        min_child_samples=30,
        subsample=0.85,
        colsample_bytree=0.85,
        reg_alpha=0.8,
        reg_lambda=1.0,
    ),
    # 7) Smaller leaves (reduces overfit; can bump MCC)
    dict(
        **base,
        num_leaves=31,
        max_depth=-1,
        min_child_samples=10,
        subsample=0.9,
        colsample_bytree=0.9,
        reg_alpha=0.0,
        reg_lambda=0.1,
    ),
    # 8) Smaller leaves + more sampling (extra robustness)
    dict(
        **base,
        num_leaves=31,
        max_depth=-1,
        min_child_samples=20,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.2,
        reg_lambda=0.5,
    ),
    # 9) Aggressive sampling (often helps public LB stability)
    dict(
        **base,
        num_leaves=63,
        max_depth=-1,
        min_child_samples=20,
        subsample=0.75,
        colsample_bytree=0.75,
        reg_alpha=0.2,
        reg_lambda=0.8,
    ),
    # 10) Slightly faster variant (higher LR fewer trees)
    dict(
        n_estimators=5000,
        learning_rate=0.02,
        random_state=BASE_SEED,
        n_jobs=-1,
        num_leaves=63,
        max_depth=-1,
        min_child_samples=15,
        subsample=0.9,
        colsample_bytree=0.9,
        reg_alpha=0.1,
        reg_lambda=0.1,
    ),
]

# ============ Run search ============
results = []
best = None

for i, params in enumerate(param_grid, 1):
    mcc, t, oof, test_proba = cv_eval_lgb_params(
        X, y, X_test, params, n_splits=N_SPLITS, seed=BASE_SEED
    )
    row = {
        "config_id": i,
        "cv_mcc": mcc,
        "best_threshold": t,
        "params": str(params),
    }
    results.append(row)
    print(f"[{i:02d}/{len(param_grid)}] CV MCC={mcc:.5f} @ t={t:.3f}")

    if best is None or mcc > best["cv_mcc"]:
        best = {
            "config_id": i,
            "cv_mcc": mcc,
            "best_threshold": t,
            "params": params,
            "test_proba": test_proba,
        }

res_df = pd.DataFrame(results).sort_values("cv_mcc", ascending=False)
res_df.to_csv("lgb_search_results.csv", index=False)

print("\nBest config:")
print(f"  config_id      : {best['config_id']}")
print(f"  CV MCC         : {best['cv_mcc']:.5f}")
print(f"  best threshold : {best['best_threshold']:.3f}")
print(f"  params         : {best['params']}")

# ============ Create best submission ============
best_pred = (best["test_proba"] >= best["best_threshold"]).astype(int)
submission = pd.DataFrame({"ID": test_ids, "target": best_pred})
submission.to_csv("results_best.csv", index=False)
print("\nSaved: results_best.csv")
print("Saved: lgb_search_results.csv")
```
