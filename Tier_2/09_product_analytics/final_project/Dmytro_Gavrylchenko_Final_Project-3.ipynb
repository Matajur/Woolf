### 1. Data loading

```python
import pandas as pd
from scipy import stats

control_df = pd.read_csv("./control_group.csv", sep=";")
test_df = pd.read_csv("./test_group.csv", sep=";")


control_df
```

```python
test_df
```

### 2. Data preparation

The control group dataset contains a day when advertising costs were recorded, but there are no statistics for this day at all, indicating a technical failure. Decision is to remove this day from the control and test groups.

```python
control_clean = control_df.dropna(subset=["# of Purchase"]).copy()
test_clean = test_df.drop(4).copy()
```

### 3. Compute key ration per day

#### 3.1. PCR (Purchase Conversion Rate)

Let's calculate the PCR two times with two different formulas basing on the Website Clicks and on the Reach.

```python
control_clean["PCR_1"] = control_clean["# of Purchase"] / control_clean["# of Website Clicks"]
test_clean["PCR_1"] = test_clean["# of Purchase"] / test_clean["# of Website Clicks"]
```

```python
control_clean["PCR_2"] = control_clean["# of Purchase"] / control_clean["Reach"]
test_clean["PCR_2"] = test_clean["# of Purchase"] / test_clean["Reach"]
```

#### 3.2. CTR (Click-Through Rate)

```python
control_clean["CTR"] = control_clean["# of Website Clicks"] / control_clean["# of Impressions"]
test_clean["CTR"] = test_clean["# of Website Clicks"] / test_clean["# of Impressions"]
```

#### 3.3. ACR (Add-to-Cart Rate)

```python
control_clean["ACR"] = control_clean["# of Add to Cart"] / control_clean["# of Website Clicks"]
test_clean["ACR"] = test_clean["# of Add to Cart"] / test_clean["# of Website Clicks"]
```

#### 3.4. CPA (Cost per Acquisition)

```python
control_clean["CPA"] = control_clean["Spend [USD]"] / control_clean["# of Purchase"]
test_clean["CPA"] = test_clean["Spend [USD]"] / test_clean["# of Purchase"]
```

```python
control_clean.head()
```

```python
test_clean.head()
```

### 4. T-test for main metrics

```python
results = {
    "PCR_1": stats.ttest_ind(control_clean["PCR_1"], test_clean["PCR_1"], equal_var=False),
    "PCR_2": stats.ttest_ind(control_clean["PCR_2"], test_clean["PCR_2"], equal_var=False),
    "CTR": stats.ttest_ind(control_clean["CTR"], test_clean["CTR"], equal_var=False),
    "ACR": stats.ttest_ind(control_clean["ACR"], test_clean["ACR"], equal_var=False),
    "CPA": stats.ttest_ind(control_clean["CPA"], test_clean["CPA"], equal_var=False),
}

results
```

### 5. Summary

```python
summary = {
    "Control Avg PCR_1": control_clean["PCR_1"].mean(),
    "Test Avg PCR_1": test_clean["PCR_1"].mean(),
    "Control Avg PCR_2": control_clean["PCR_2"].mean(),
    "Test Avg PCR_2": test_clean["PCR_2"].mean(),
    "Control Avg CTR": control_clean["CTR"].mean(),
    "Test Avg CTR": test_clean["CTR"].mean(),
    "Control Avg ACR": control_clean["ACR"].mean(),
    "Test Avg ACR": test_clean["ACR"].mean(),
    "Control Avg CPA": control_clean["CPA"].mean(),
    "Test Avg CPA": test_clean["CPA"].mean(),
}

summary
```

### 6. Conclusion

- PCR (Purchase Conversion Rate)

    First method:
    * Control: 11%;
    * Test: 9%;
    * The test group gets 1.2x lower PCR;
    * t = 1.6;
    * p = 0.1232 > 0.05;
    * Statistically not significant difference: control group is slightly better.

    Alternative method:
    * Control: 0.6%;
    * Test: 1.4%;
    * The test group gets 2.3x higher PCR;
    * t = -3;
    * p = 0.0045 < 0.05;
    * Highly significant improvement in the primary metric, which reflects an increase in purchases depending on reach.

- CTR (Click-Through Rate)
    * Control: 5%;
    * Test: 10%;
    * The test group gets 2x higher CTR;
    * t = -4;
    * p = 0.0003 < 0.05;
    * Highly significant improvement in the click rate on the new ad.

- ACR (Add-to-Cart Rate)
    * Control: 28%;
    * Test: 16%;
    * Despite a significant increase in the number of users attracted to the website by the new ad (CTR), the test group gets 1.75x lover ACR, so smaller percentage of potential customers have the desire to make a purchase;
    * t = 3.8;
    * p = 0.0005 < 0.05;
    * Highly significant deterioration in the desire to make a purchase.

- CPA (Cost per Acquisition)
    * Control: 5.1 USD;
    * Test: 6.0 USD;
    * The test group has 1.2x higher cost of acquiring a new customer;
    * t = -1.5;
    * p = 0.1516 > 0.05;
    * Not significant difference: the test group worse on average, but not statistically proven.

The results cannot be interpreted unambiguously. The test campaign has a much higher CTR (click-through rate), that means more people move from Reach to Website Click. Even if the on-site conversion is slightly worse, the larger volume of visitors generates more purchases relative to total reach. But a significant decrease in Add-to-Cart Rate may indicate that the site itself or the products offered do not match what potential customers imagine after viewing the ad, so they may feel cheated, o something wrong with the landing page. More research on this topic is needed.

Depending on the business strategy, the following options can be offered:
- If the business goal is maximize purchases per impression/reach, the test campaign is clearly better, because it brings in more buyers per person reached.
- If the goal is improve website UX and efficiency, the control looks healthier — meaning the landing page/product experience for test traffic or the ad may need optimization.
