### 1. Data loading

```python
import pandas as pd
import numpy as np
from math import sqrt
from scipy import stats

control_df = pd.read_csv("./control_group.csv", sep=";")
test_df = pd.read_csv("./test_group.csv", sep=";")


control_df.head()
```

```python
test_df.head()
```

### 2. Data preparation

The control group dataset contains a day when advertising costs were recorded, but there are no statistics for this day at all, indicating a technical failure. Decision is to remove this day from the control and test groups.

```python
control_clean = control_df.dropna(subset=["# of Purchase"]).copy()
test_clean = test_df.drop(4).copy()
```

### 3. Data aggregation

```python
control_atc = int(control_df["# of Add to Cart"].sum())
control_buy = int(control_df["# of Purchase"].sum())
test_atc = int(test_df["# of Add to Cart"].sum())
test_buy = int(test_df["# of Purchase"].sum())
```

### 4. Metrics calculation

Delta conversion

```python
conversion_test = test_buy / test_atc
conversion_control = control_buy / control_atc

conversion_test, conversion_control
```

```python
delta = conversion_test - conversion_control
delta
```

Standard error for difference in proportions

```python
se = sqrt(conversion_test * (1 - conversion_test) / test_atc + conversion_control * (1 - conversion_control) / control_atc)
se
```

95% confidence interval

```python
z = stats.norm.ppf(0.975)
ci_low = delta - z * se
ci_high = delta + z * se

ci_low, ci_high
```

### 5. Summary

From the previous task, it is clear that the “Add-to-Cart → Purchase” conversion for the test group is 59.13%, and for the control group it is 40.21%. The difference is 18.91%. 95% CI is from 18.14% to 19.69%. Std error = 0.0039.
