```python
name: CI
on: [push]

jobs:
  secrets:
    steps:
      - uses: actions/checkout@v3
      - name: Gitleaks scan
        run: gitleaks detect --exit-code 1

  sbom:
    steps:
      - run: syft . -o cyclonedx-json > sbom.json
      - run: trivy sbom sbom.json --exit-code 1

  build:
    runs-on: ubuntu-latest
    steps:
      - name: Checkout
        uses: actions/checkout@v3

      - name: Build container
        run: docker build -t user-api .
```
